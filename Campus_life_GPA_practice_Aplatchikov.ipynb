```python
y = df['gpa'].copy()
numeric_features = ['student_year',
                    'sleep_hours_avg',
                    'study_hours_week',
                    'library_minutes_week',
                    'phone_screen_hours_day',
                    'cafe_visits_week', 'coffee_cups_week',
                    'stress_1_10', 'steps_avg_day',
                    'commute_minutes',
                    'part_time_hours_week',
                    'group_study_sessions']

X = pd.get_dummies(df[numeric_features + ['faculty']],
                   columns = ['faculty'])


X.head()
```

Train/test split

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.25,random_state=42
    )
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)

pred = lr.predict(X_test)

print('mse:',round(mean_squared_error(y_test,pred),4))
print('r2 score', round(r2_score(y_test,pred),4))
```

Задания на самостоятельную работу

Scaling

1. Добавить Scaling и сравнить r2_score с и без scaling'a

```python
scaler = StandardScaler()

X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.fit_transform(X_test)

lr = LinearRegression()
lr.fit(X_train_s,y_train)

pred_sc = lr.predict(X_test_s)

print('mse scaled:',round(mean_squared_error(y_test,pred_sc),4))
print('r2 score scaled', round(r2_score(y_test,pred_sc),4))

print("difference = ",round(r2_score(y_test,pred_sc),4) -  round(r2_score(y_test,pred),4))


```

 Сравнить Ridge vs LinearRegression

```python
best_r2 = 0
best_a = 0

for i in range(1,100):
  lr_r = Ridge(alpha = i/100)
  lr_r.fit(X_train_s,y_train)

  if r2_score(y_test,pred) > best_r2:
    best_r2 = r2_score(y_test,pred)
    best_a = i/100

  pred_r = lr_r.predict(X_test_s)


print("best a", best_a, "r2", best_r2)

print("difference", best_r2 - r2_score(y_test,pred_sc))
```
